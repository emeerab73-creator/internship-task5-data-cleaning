```python
import pandas as pd
import numpy as np
import random

random.seed(42)
np.random.seed(42)

first_names = ["Ayesha","Bilal","Sana","Hamza","Zainab","Ali","Mahnoor","Usman","Fatima","Hassan",
               "Iqra","Ahmed","Sadia","Fahad","Mariam","Umar","Rabia","Zeeshan","Nida","Talha"]
last_names = ["Khan","Malik","Ahmed","Iqbal","Chaudhry","Butt","Raza","Hussain","Farooq","Sheikh"]

city_variants = {
    "Lahore": ["Lahore","lahore","LAHORE","Lahore ","Lahor"],
    "Karachi": ["Karachi","karachi","KARACHI","Karachi "],
    "Islamabad": ["Islamabad","islamabad","ISLAMABAD"],
    "Multan": ["Multan","multan"],
}
cities_clean = list(city_variants.keys())
universities = ["University of Education Lahore","LUMS","FAST-NUCES","Punjab University","COMSATS"]
degrees = ["BS English","BS Computer Science","BBA","BS Software Engineering","BS Mass Communication"]
skills_pool = ["Excel","Python","Canva","Content Writing","SEO","Social Media Marketing",
               "Graphic Design","Data Analysis","UI/UX Design","MS Office","Photoshop"]
roles = ["Digital Marketing Intern","Content Writing Intern","Graphic Design Intern",
         "Data Analytics Intern","UI/UX Design Intern","Virtual Assistant Intern"]
statuses = ["Applied","Shortlisted","Rejected","Selected"]

rows = []
for i in range(1, 121):
    fname, lname = random.choice(first_names), random.choice(last_names)
    age = int(np.random.normal(22, 2))
    city_clean = random.choice(cities_clean)
    city = random.choice(city_variants[city_clean])
    skills_str = ", ".join(random.sample(skills_pool, k=random.randint(2,4)))
    date = pd.Timestamp('2024-01-01') + pd.Timedelta(days=random.randint(0,150))
    date_str = date.strftime(random.choice(["%Y-%m-%d","%d/%m/%Y","%B %d, %Y"]))
    email = f"{fname.lower()}.{lname.lower()}{random.randint(1,99)}@gmail.com"
    rows.append([i, f"{fname} {lname}", age, city, random.choice(universities),
                 random.choice(degrees), skills_str, random.choice(roles),
                 date_str, random.choice(statuses), email])

df = pd.DataFrame(rows, columns=["applicant_id","full_name","age","city","university",
                                   "degree","skills","applied_role","application_date","status","email"])

for col in ["age","city","skills","email"]:
    df.loc[df.sample(frac=0.06).index, col] = np.nan

df.loc[df.sample(4).index, "age"] = [150, -5, 3, 99]
df = pd.concat([df, df.sample(6)], ignore_index=True)

df.head()
```
